# Setup

```python
import sys
import os
import pandas as pd
from IPython.display import display

# set path
project_root = os.path.abspath('')
while not os.path.isdir(os.path.join(project_root, 'utilities')):
    parent_dir = os.path.dirname(project_root)
    project_root = parent_dir

if project_root not in sys.path:
    sys.path.insert(0, project_root)
print(f"Project root set to: {project_root}")

# import testing modules
from utilities.hypothesis_testing.main import main as run_pipeline
```

# Load data

```python
dataset_filename = "ocean_current.csv" 

run_ks = True
run_bivariate = True
run_independence = True
graph_types_to_test = ['SmallWorld', 'StochasticBlock', 'ScaleFree']

base_dataset_name = dataset_filename.split('.')[0]
data_dir = os.path.join(project_root, 'data', 'real_world', 'preprocessed')
input_path = os.path.join(data_dir, dataset_filename)
```

# Run tests

```python
for g_type in graph_types_to_test:

    print(f"STARTING TEST FOR: {g_type.upper()} CLASSIFICATION")

    output_dir = os.path.join(project_root, 'results', 'real_world') 
    batch_id = base_dataset_name
    dataset_id = g_type
    
    os.makedirs(output_dir, exist_ok=True)
    

    run_pipeline(
        input_path=input_path,
        dataset_id=dataset_id,
        batch_id=batch_id,
        output_dir=output_dir,
        run_ks=run_ks,
        run_independence=run_independence,
        run_bivariate=run_bivariate,
        jobs=-1,
        graph_type=g_type,
        data_type='real',
        auto_tune=True,
        apply_jitter=False
    )
    report_csv_path = os.path.join(output_dir, batch_id, "summary_report.csv")
    
    if os.path.exists(report_csv_path):
            df_report = pd.read_csv(report_csv_path)
            current_report = df_report[df_report['Dataset ID'] == g_type]
            display(current_report)
```
